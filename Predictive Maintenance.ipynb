```python
# Select sensor and operating-setting features
feature_columns = [
    col for col in df.columns
    if col.startswith("sensor_") or col.startswith("setting_")
]

X = df[feature_columns]
y = df["RUL"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Number of features:", len(feature_columns))
print("Missing values in X:", X.isnull().sum().sum())
print("Missing values in y:", y.isnull().sum())
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Keep all rows from each engine in the same split
groups = df["unit_number"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

train_engines = set(groups.iloc[train_idx])
test_engines = set(groups.iloc[test_idx])

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training engines:", len(train_engines))
print("Testing engines:", len(test_engines))
print("Shared engines:", len(train_engines & test_engines))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Create the model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train the model
model.fit(X_train, y_train)

# Predict RUL for unseen engines
y_pred = model.predict(X_test)

# Evaluate predictions
mae = mean_absolute_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Create a simple baseline model
baseline = DummyRegressor(strategy="mean")

# Train using the same training data
baseline.fit(X_train, y_train)

# Predict on the same test engines
baseline_pred = baseline.predict(X_test)

# Evaluate baseline
baseline_mae = mean_absolute_error(y_test, baseline_pred)
baseline_rmse = root_mean_squared_error(y_test, baseline_pred)

print("Baseline MAE:", round(baseline_mae, 2))
print("Baseline RMSE:", round(baseline_rmse, 2))

print("\nRandom Forest MAE:", round(mae, 2))
print("Random Forest RMSE:", round(rmse, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Cap RUL at 125 cycles
y_capped = df["RUL"].clip(upper=125)

# Use the SAME engine-based train/test split
y_train_capped = y_capped.iloc[train_idx]
y_test_capped = y_capped.iloc[test_idx]

# Train Random Forest on the capped target
rf_capped = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_capped.fit(X_train, y_train_capped)
pred_capped = rf_capped.predict(X_test)

# Train a baseline on the same capped target
baseline_capped = DummyRegressor(strategy="mean")
baseline_capped.fit(X_train, y_train_capped)
baseline_pred_capped = baseline_capped.predict(X_test)

# Evaluate both models against the same target
print("Capped Random Forest")
print("MAE:", round(mean_absolute_error(y_test_capped, pred_capped), 2))
print("RMSE:", round(root_mean_squared_error(y_test_capped, pred_capped), 2))

print("\nCapped Baseline")
print("MAE:", round(mean_absolute_error(y_test_capped, baseline_pred_capped), 2))
print("RMSE:", round(root_mean_squared_error(y_test_capped, baseline_pred_capped), 2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test_capped, pred_capped, alpha=0.3)

plt.plot([0, 125], [0, 125], "r--")

plt.xlabel("Actual RUL (capped at 125)")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted RUL - Random Forest")
plt.xlim(0, 125)
plt.ylim(0, 125)
plt.grid(True)
plt.show()
```

```python
# Create a table of actual and predicted RUL
results = pd.DataFrame({
    "Actual_RUL": y_test_capped.to_numpy(),
    "Predicted_RUL": pred_capped
})

# Calculate prediction errors
results["Absolute_Error"] = (
    results["Actual_RUL"] - results["Predicted_RUL"]
).abs()

# Group test samples by actual RUL
results["Life_Stage"] = np.where(
    results["Actual_RUL"] <= 30,
    "Critical (0-30)",
    np.where(
        results["Actual_RUL"] <= 75,
        "Middle (31-75)",
        "Early (76-125)"
    )
)

# Evaluate errors in each stage
stage_results = results.groupby("Life_Stage").agg(
    Samples=("Actual_RUL", "count"),
    MAE=("Absolute_Error", "mean")
)

display(stage_results.round(2))
```

```python
results["Error"] = (
    results["Predicted_RUL"] - results["Actual_RUL"]
)

error_analysis = results.groupby("Life_Stage").agg(
    Samples=("Error", "count"),
    Mean_Error=("Error", "mean"),
    MAE=("Absolute_Error", "mean")
)

display(error_analysis.round(2))
```

```python
from sklearn.model_selection import GroupShuffleSplit

groups = df["unit_number"]

# 20% للاختبار النهائي
splitter_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_val_idx, test_idx = next(
    splitter_test.split(X, y_capped, groups=groups)
)

# تقسيم الجزء المتبقي إلى تدريب وتحقق
X_train_val = X.iloc[train_val_idx]
y_train_val = y_capped.iloc[train_val_idx]
groups_train_val = groups.iloc[train_val_idx]

splitter_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx2, val_idx2 = next(
    splitter_val.split(
        X_train_val,
        y_train_val,
        groups=groups_train_val
    )
)

X_train_final = X_train_val.iloc[train_idx2]
y_train_final = y_train_val.iloc[train_idx2]

X_val = X_train_val.iloc[val_idx2]
y_val = y_train_val.iloc[val_idx2]

X_test_final = X.iloc[test_idx]
y_test_final = y_capped.iloc[test_idx]

print("Training rows:", len(X_train_final))
print("Validation rows:", len(X_val))
print("Testing rows:", len(X_test_final))

print(
    "Shared engines between train and validation:",
    len(
        set(groups.iloc[train_idx].tolist())
        & set(groups.iloc[test_idx].tolist())
    )
)
```

```python
train_engines_final = set(
    groups_train_val.iloc[train_idx2]
)

val_engines_final = set(
    groups_train_val.iloc[val_idx2]
)

test_engines_final = set(
    groups.iloc[test_idx]
)

print("Train-Val overlap:", len(train_engines_final & val_engines_final))
print("Train-Test overlap:", len(train_engines_final & test_engines_final))
print("Val-Test overlap:", len(val_engines_final & test_engines_final))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    )
}

validation_results = []

for name, model in models.items():
    model.fit(X_train_final, y_train_final)

    val_pred = model.predict(X_val)

    mae = mean_absolute_error(y_val, val_pred)
    rmse = np.sqrt(mean_squared_error(y_val, val_pred))

    validation_results.append({
        "Model": name,
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2)
    })

validation_df = pd.DataFrame(validation_results)

display(validation_df.sort_values("MAE"))
```
